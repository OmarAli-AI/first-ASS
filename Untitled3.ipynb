```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures

data = pd.read_csv('Ice_cream selling data.csv')
X_input = data[['Temperature (°C)']].values
y_target = data['Ice Cream Sales (units)'].values


poly_feat = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_feat.fit_transform(X_input)



X_mean = np.mean(X_poly, axis=0)
X_std = np.std(X_poly, axis=0)
X_poly = (X_poly - X_mean) / X_std

y_mean = np.mean(y_target)
y_std = np.std(y_target)
y_scaled = (y_target - y_mean) / y_std


np.random.seed(42)
weights = np.zeros(X_poly.shape[1])
bias = 0.0

learning_rate = 0.01
momentum_rate = 0.9
n_epochs = 50
m_samples = len(y_scaled)

v_weights = np.zeros_like(weights)
v_bias = 0.0

for epoch in range(n_epochs):


    shuffled_idx = np.random.permutation(m_samples)
    X_shuffled = X_poly[shuffled_idx]
    y_shuffled = y_scaled[shuffled_idx]



    for i in range(m_samples):
        xi = X_shuffled[i:i+1]
        yi = y_shuffled[i:i+1]

        y_predicted = np.dot(xi, weights) + bias
        err = y_predicted - yi

        grad_w = xi.T * err
        grad_b = err[0]


        v_weights = momentum_rate * v_weights + learning_rate * grad_w.flatten()
        v_bias = momentum_rate * v_bias + learning_rate * grad_b


        weights = weights - v_weights
        bias = bias - v_bias

print("SGD Training finished successfully ")



plt.figure(figsize=(8, 5))
plt.scatter(X_input, y_scaled, color='blue', label='Original Data')
y_final_pred = np.dot(X_poly, weights) + bias
sorted_indices = np.argsort(X_input.flatten())
plt.plot(X_input[sorted_indices], y_final_pred[sorted_indices], color='green', linewidth=2, label='SGD Momentum Fit')
plt.xlabel('Temperature')
plt.ylabel('Sales')
plt.title('Stochastic Gradient Descent (SGD) with Momentum')
plt.legend()
plt.show()
```
