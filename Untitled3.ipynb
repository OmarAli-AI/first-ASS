```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures

# تحميل البيانات (تأكد أن ملف الـ CSV مرفوع في نافذة الملفات على اليسار)
df = pd.read_csv('Ice_cream selling data.csv')

# استخراج المتغيرات
X = df[['Temperature (°C)']].values
y = df['Ice Cream Sales (units)'].values

# تحويل البيانات إلى Polynomial (Degree = 2)
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

# توحيد المقاييس (Feature Scaling) لتسريع وتسهيل التدريب
X_mean = np.mean(X_poly, axis=0)
X_std = np.std(X_poly, axis=0)
X_poly = (X_poly - X_mean) / X_std

y_mean = np.mean(y)
y_std = np.std(y)
y_scaled = (y - y_mean) / y_std

print("تم تجهيز البيانات بنجاح!")
```

```python
# تهيئة الأوزان والـ Bias
np.random.seed(42)
w = np.zeros(X_poly.shape[1])
b = 0.0

# الـ Hyperparameters
lr = 0.05
beta = 0.9          # معامل الزخم (Momentum)
epochs = 1000
m = len(y_scaled)

# متغيرات لحفظ الخطأ لكل Epoch
cost_history = []

# المتجهات الخاصة بالزخم (Velocity)
v_w = np.zeros_like(w)
v_b = 0.0

for epoch in range(epochs):
    # 1. التنبؤ
    y_pred = np.dot(X_poly, w) + b

    # 2. حساب الخطأ (Cost - MSE)
    error = y_pred - y_scaled
    cost = (1 / (2 * m)) * np.sum(error ** 2)
    cost_history.append(cost)

    # 3. حساب التدرج (Gradients)
    dw = (1 / m) * np.dot(X_poly.T, error)
    db = (1 / m) * np.sum(error)

    # 4. تحديث سرعة الزخم (Velocity Update)
    v_w = beta * v_w + lr * dw
    v_b = beta * v_b + lr * db

    # 5. تحديث الأوزان الفعلية
    w = w - v_w
    b = b - v_b

print(f"انتهى التدريب! الخسارة النهائية: {cost_history[-1]:.5f}")
```

```python
# رسم منحنى التنبؤ مقابل البيانات الحقيقية
plt.figure(figsize=(8, 5))
plt.scatter(X, y_scaled, color='blue', label='Actual Data')
y_pred_final = np.dot(X_poly, w) + b
sort_idx = np.argsort(X.flatten())
plt.plot(X[sort_idx], y_pred_final[sort_idx], color='red', linewidth=2, label='Polynomial + Momentum GD')
plt.xlabel('Temperature (°C)')
plt.ylabel('Scaled Sales')
plt.title('Batch Gradient Descent with Momentum - Polynomial Regression')
plt.legend()
plt.show()
```
